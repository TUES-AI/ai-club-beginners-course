## Задача 1 ⭐ Вдигнат ли е пръстът
Напиши функция `is_up(ys, tip, joint)`, която връща `True`, ако върхът на пръста (точка `tip`) е по-високо от ставата му (точка `joint`).

**Как мислим**
- Кога пръстът е вдигнат? Когато върхът му е над ставата.
- Кое число казва колко високо е една точка? `ys` с нейния номер. По-високо значи по-малко `y`.
- Сравнението `ys[tip] < ys[joint]` вече е `True` или `False`, затова го връщаме направо.

**Пишем по стъпки:** първо отпечатайте `paper_ys[8]` и `paper_ys[6]`, за да видят числата. После функцията на един ред.

````python
def is_up(ys, tip, joint):
    return ys[tip] < ys[joint]


print(is_up(paper_ys, 8, 6))       # True
print(is_up(rock_ys, 8, 6))        # False
print(is_up(scissors_ys, 12, 10))  # True
print(is_up(scissors_ys, 16, 14))  # False
````

**Чести грешки:** знакът е обърнат (`>`), защото очакват по-голямото число да е по-високо; разменени `tip` и `joint` при извикването.

## Задача 2 ⭐ Колко пръста са вдигнати
Напиши функция `count_fingers(ys)`, която връща колко от четирите пръста (без палеца) са вдигнати. Използвай `is_up`.

**Как мислим**
- Имаме четири пръста и за всеки трябва да извикаме `is_up` с неговия връх и става.
- Номерата ги слагаме в два списъка: `tips` и `joints`. Както `weights` и `labels` в урок 3: `tips[i]` и `joints[i]` са на един и същ пръст.
- Броим с шаблона „брояч“.

**Пишем по стъпки:** двата списъка → `count = 0` → `for i in range(4)` → `if is_up(...)` → `return count` след цикъла.

````python
def count_fingers(ys):
    tips = [8, 12, 16, 20]
    joints = [6, 10, 14, 18]
    count = 0
    for i in range(4):
        if is_up(ys, tips[i], joints[i]):
            count += 1
    return count


print(count_fingers(rock_ys))      # 0
print(count_fingers(scissors_ys))  # 2
print(count_fingers(paper_ys))     # 4
print(count_fingers(three_ys))     # 3
````

**Чести грешки:** `range(5)` дава `IndexError`; `return count` вътре в цикъла връща 0 или 1.

Вариант без списъци, с четири отделни `if`, също е верен. Покажи го, ако някой го е написал така, и сравнете двата.

## Задача 3 ⭐ Кой е жестът
Напиши функция `gesture(ys)`, която връща `"камък"` при 0 вдигнати пръста, `"ножица"` при 2, `"хартия"` при 4 и `"не знам"` във всички останали случаи.

**Как мислим**
- Броя на пръстите го взимаме от `count_fingers` и го пазим в променлива, за да не го смятаме няколко пъти.
- Три `if` с `return`. Ако никой не се изпълни, стигаме до последния ред.

**Пишем по стъпки:** `n = count_fingers(ys)` → трите условия → `return "не знам"`.

````python
def gesture(ys):
    n = count_fingers(ys)
    if n == 0:
        return "камък"
    if n == 2:
        return "ножица"
    if n == 4:
        return "хартия"
    return "не знам"


print(gesture(rock_ys))      # камък
print(gesture(scissors_ys))  # ножица
print(gesture(paper_ys))     # хартия
print(gesture(three_ys))     # не знам
print(gesture(horns_ys))     # ножица
````

**За обсъждане:** `horns_ys` е показалец и кутре, а функцията казва „ножица“. Правилото брои пръстите, но не гледа кои са. Задача 7 го поправя.

## Задача 4 ⭐⭐ Кой печели
Напиши функция `winner(me, computer)`. Двата параметъра са текст: `"камък"`, `"ножица"` или `"хартия"`. Функцията връща `"ти"`, `"компютър"` или `"равни"`.

Камъкът бие ножицата, ножицата бие хартията, хартията бие камъка.

**Как мислим**
- Кой е най-лесният случай? Двата хода са еднакви.
- После изброяваме трите случая, в които печелим ние. Всяко условие има две части, свързани с `and`.
- Ако не сме спечелили и не са равни, печели компютърът. За него не ни трябват условия.

**Пишем по стъпки:** равни → трите победи → `return "компютър"`.

````python
def winner(me, computer):
    if me == computer:
        return "равни"
    if me == "камък" and computer == "ножица":
        return "ти"
    if me == "ножица" and computer == "хартия":
        return "ти"
    if me == "хартия" and computer == "камък":
        return "ти"
    return "компютър"


print(winner("камък", "ножица"))   # ти
print(winner("камък", "хартия"))   # компютър
print(winner("хартия", "хартия"))  # равни
print(winner("ножица", "хартия"))  # ти
print(winner("ножица", "камък"))   # компютър
````

Трите победи могат да се съберат в едно условие с `or`. Покажи го, ако някой попита:

```python
if (me == "камък" and computer == "ножица") or (me == "ножица" and computer == "хартия") or (me == "хартия" and computer == "камък"):
    return "ти"
```

**Чести грешки:** правописна разлика в текстовете, например `"Камък"` с главна буква. Сравнението на текст е точно.

## Задача 5 ⭐⭐ Резултат от няколко игри
Изиграни са няколко игри. Ходовете са в два списъка: `my_moves[i]` е твоят ход в игра `i`, а `computer_moves[i]` е ходът на компютъра.

Напиши функция `score(my_moves, computer_moves)`, която връща колко игри си спечелил. Използвай `winner`.

**Как мислим**
- Пак два списъка, в които един и същ индекс значи една и съща игра.
- За всяка игра питаме `winner` и броим само случаите, в които отговорът е `"ти"`.

**Пишем по стъпки:** `wins = 0` → `for i in range(len(my_moves))` → `if winner(...) == "ти"` → `return wins`.

````python
def score(my_moves, computer_moves):
    wins = 0
    for i in range(len(my_moves)):
        if winner(my_moves[i], computer_moves[i]) == "ти":
            wins += 1
    return wins


mine = ["камък", "хартия", "ножица", "камък", "хартия"]
comp = ["ножица", "хартия", "камък", "хартия", "камък"]
print(score(mine, comp))            # 2
print(score(["камък"], ["камък"]))  # 0
````

## Задача 6 ⭐⭐ Вляво или вдясно е ръката
Списъкът `xs` казва колко наляво или надясно е всяка точка: 0 е левият край на картината, 100 е десният.

Напиши функция `side(xs)`, която връща `"ляво"`, ако **средното** на всички числа в `xs` е под 50, и `"дясно"` иначе.

**Как мислим**
- Една точка не стига, защото ръката е широка. Взимаме средното на всички 21 точки.
- Средно на списък сме смятали в урок 3: `sum(xs) / len(xs)`.

**Пишем по стъпки:** `avg = sum(xs) / len(xs)` → `if avg < 50` → двата `return`.

````python
def side(xs):
    avg = sum(xs) / len(xs)
    if avg < 50:
        return "ляво"
    return "дясно"


print(side(far_xs))    # ляво
print(side(right_xs))  # дясно
````

## Задача 7 ⭐⭐⭐ Истинска ножица
Функцията `gesture` казва „ножица“ за всеки два вдигнати пръста. Искаме да е по-точна.

1. Напиши функция `raised(ys)`, която връща **списък с имената** на вдигнатите пръсти, в реда показалец, среден, безименен, кутре.
2. Напиши функция `is_scissors(ys)`, която връща `True` само ако са вдигнати точно показалецът и средният пръст. Използвай `raised`.

*Подсказка:* два списъка могат да се сравнят с `==`. Равни са, ако имат едни и същи елементи в един и същ ред.

**Как мислим**
- `raised` е `count_fingers`, но вместо да броим, добавяме името на пръста в списък. Трябва ни трети списък с имената.
- Това е шаблонът от урок 3: празен списък, `append` в цикъла, `return` след него.
- За `is_scissors` сравняваме получения списък с този, който искаме.

**Пишем по стъпки:** копирайте `count_fingers` → добавете `names` → сменете брояча със списък → `is_scissors` на един ред.

````python
def raised(ys):
    tips = [8, 12, 16, 20]
    joints = [6, 10, 14, 18]
    names = ["показалец", "среден", "безименен", "кутре"]
    result = []
    for i in range(4):
        if is_up(ys, tips[i], joints[i]):
            result.append(names[i])
    return result


def is_scissors(ys):
    return raised(ys) == ["показалец", "среден"]


print(raised(scissors_ys))       # ['показалец', 'среден']
print(raised(horns_ys))          # ['показалец', 'кутре']
print(raised(rock_ys))           # []
print(is_scissors(scissors_ys))  # True
print(is_scissors(horns_ys))     # False
print(is_scissors(paper_ys))     # False
````

    return is_up(ys, 8, 6) and is_up(ys, 12, 10) and not is_up(ys, 16, 14) and not is_up(ys, 20, 18)

## Задача 8 ⭐⭐⭐ Близо или далеч е ръката
Когато ръката е близо до камерата, тя изглежда по-голяма. Големината ѝ можем да измерим с дължината на дланта: от китката (точка 0) до основата на средния пръст (точка 9).

1. Напиши функция `palm(ys)`, която връща с колко точка 0 е по-ниско от точка 9.
2. Напиши функция `is_close(ys)`, която връща `True`, ако дланта е по-дълга от 30.

**Как мислим**
- Китката е долу, значи нейното `y` е по-голямо. Дължината е `ys[0] - ys[9]`.
- Защо мерим дланта, а не цялата ръка от най-ниската до най-високата точка? Защото при юмрук ръката е по-ниска, без да е по-далеч. Дланта не се променя, когато свиваме пръстите.

**Пишем по стъпки:** `palm` на един ред → отпечатайте я за `paper_ys`, `rock_ys` и `far_ys` → изберете заедно прага → `is_close`.

````python
def palm(ys):
    return ys[0] - ys[9]


def is_close(ys):
    return palm(ys) > 30


print(palm(paper_ys))     # 38
print(palm(far_ys))       # 20
print(is_close(paper_ys)) # True
print(is_close(rock_ys))  # True
print(is_close(far_ys))   # False
````

**За обсъждане:** това е първото „разстояние“ между две точки, но само по височина. Ако ръката е наклонена, сметката вече не е точна. Следващия път смятаме истинското разстояние между две точки.
