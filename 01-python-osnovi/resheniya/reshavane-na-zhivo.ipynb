## Задача 1 ⭐ Най-голямото от N числа
Въведи **N**, после N цели числа. Отпечатай най-голямото от тях.

| Вход | Изход |
|---|---|
| `5`, после `3 9 -2 7 4` (по едно на ред) | `9` |

**Как мислим**
- Ако имаш купчина карти с числа и ги гледаш една по една — как намираш най-голямата? Държиш „най-голямата досега“ и сменяш, ако видиш по-голяма.
- С какво да започнем? Питай: защо не с `best = 0`? → ако всички числа са отрицателни, ще отпечатаме 0, а го няма!

**Пишем по стъпки**
1. `n = int(input(...))`
2. Първото число отделно → то е `best`.
3. Цикъл за останалите `n - 1` числа: въвеждаме, сравняваме.
4. `print` след цикъла.

````python
n = int(input("N: "))
best = int(input("Число: "))
for i in range(n - 1):
    x = int(input("Число: "))
    if x > best:
        best = x
print(best)
````

**Проверка:** `5 → 3 9 -2 7 4` → `9` · `3 → -5 -2 -8` → `-2`

**Чести грешки:** `best = 0` в началото; `range(n)` вместо `range(n - 1)` (иска едно число повече).

## Задача 2 ⭐ Брой на цифрите
Въведи естествено число **N**. Отпечатай колко цифри има.

| Вход | Изход |
|---|---|
| `2026` | `4` |
| `7` | `1` |

**Как мислим**
- Колко пъти можем да „отрежем“ последната цифра (`// 10`), докато числото стане 0? Точно толкова цифри има.
- Не знаем колко цифри има предварително → `while`.

**Пишем по стъпки**
1. Вход.
2. `count = 0` (шаблон брояч).
3. `while n > 0:` → `count += 1`, `n = n // 10`.
4. `print(count)`.

````python
n = int(input("N: "))
count = 0
while n > 0:
    count += 1
    n = n // 10
print(count)
````

**Проверка:** `2026` → `4` · `7` → `1` · `1000` → `4`

**За любопитните:** в Python може и `len(str(n))` — но в училище искат с цикъл, а и така разбираме как работи.

## Задача 3 ⭐ Сума на цифрите
Въведи естествено число **N**. Отпечатай сбора на цифрите му.

| Вход | Изход |
|---|---|
| `2026` | `10` |

**Как мислим**
- Копирай задача 2 — какво сменяме? Вместо `count += 1` → `total += n % 10`.
- Покажи им, че почти всички задачи с цифри са **същият цикъл** с различна среда.

**Пишем по стъпки**
1. Вход, `total = 0`.
2. Цикълът за цифри: `digit = n % 10`, `total += digit`, `n = n // 10`.
3. `print(total)`.

````python
n = int(input("N: "))
total = 0
while n > 0:
    digit = n % 10
    total += digit
    n = n // 10
print(total)
````

**Проверка:** `2026` → `10` · `999` → `27`

**Чести грешки:** `total += n` вместо `total += n % 10`.

## Задача 4 ⭐⭐ Числата на Фибоначи
Първите две числа на Фибоначи са **0** и **1**, а всяко следващо е сборът на предходните две. Въведи **N** и отпечатай първите N числа на един ред.

| Вход | Изход |
|---|---|
| `8` | `0 1 1 2 3 5 8 13` |

**Как мислим**
- Напишете на дъската: 0, 1, 1, 2, 3, 5, 8… Какво ни трябва, за да получим следващото? Само **последните две**.
- Пазим ги в `a` и `b`. След всяка стъпка: новото `a` е старото `b`, новото `b` е `a + b`. Това е размяната от презентацията!

**Пишем по стъпки**
1. Вход, `a = 0`, `b = 1`.
2. `for i in range(n):` → `print(a, end=" ")`.
3. `a, b = b, a + b`.
4. Покажи и версията с `temp` (по-долу) — така се прави на C/C++.

````python
n = int(input("N: "))
a = 0
b = 1
for i in range(n):
    print(a, end=" ")
    a, b = b, a + b
print()
````

Същото с трета променлива (както на други езици):

```python
for i in range(n):
    print(a, end=" ")
    temp = a + b
    a = b
    b = temp
```

**Проверка:** `8` → `0 1 1 2 3 5 8 13` · `1` → `0`

**Чести грешки:** `a = b` и после `b = a + b` на два реда — `a` вече е сменено и излиза грешна редица. (Пусни го — добър момент за учене!)

## Задача 5 ⭐⭐ Делители и техният брой
Въведи естествено число **N**. Отпечатай всичките му делители на един ред и колко са.

| Вход | Изход |
|---|---|
| `12` | `1 2 3 4 6 12` <br> `Брой: 6` |

**Как мислим**
- Кои числа могат да са делители на 12? Само от 1 до 12 → `for d in range(1, n + 1)`.
- Как разбираме дали `d` дели `n`? Остатъкът е 0 → `n % d == 0`.
- Два шаблона наведнъж: отпечатваме **и** броим.

**Пишем по стъпки**
1. Вход, `count = 0`.
2. `for d in range(1, n + 1):` с `if n % d == 0:`.
3. Вътре: `print(d, end=" ")` и `count += 1`.
4. След цикъла: `print()` (нов ред) и `print("Брой:", count)`.

````python
n = int(input("N: "))
count = 0
for d in range(1, n + 1):
    if n % d == 0:
        print(d, end=" ")
        count += 1
print()
print("Брой:", count)
````

**Проверка:** `12` → `1 2 3 4 6 12`, `Брой: 6` · `7` → `1 7`, `Брой: 2`

**Чести грешки:** `range(1, n)` — пропуска самото N; `range(0, ...)` — деление на нула (`ZeroDivisionError`).

## Задача 6 ⭐⭐ Числото наобратно
Въведи естествено число **N**. Получи **числото** M, записано с цифрите на N наобратно, и го отпечатай.

| Вход | Изход |
|---|---|
| `1234` | `4321` |
| `120` | `21` |

**Как мислим**
- Цифрите излизат отзад напред (4, 3, 2, 1) — точно в реда, който ни трябва!
- Как „залепваме“ цифра отдясно на число? `m * 10 + цифра`. Пример на дъската: 43 → 43·10 + 2 = 432.
- Проследете с таблица: `m`: 0 → 4 → 43 → 432 → 4321.

**Пишем по стъпки**
1. Вход, `m = 0`.
2. Цикълът за цифри: `m = m * 10 + n % 10`, `n = n // 10`.
3. `print(m)`.

````python
n = int(input("N: "))
m = 0
while n > 0:
    m = m * 10 + n % 10
    n = n // 10
print(m)
````

**Проверка:** `1234` → `4321` · `120` → `21` (водещата нула изчезва — числото е 21)

**Чести грешки:** да печатат цифрите една по една с `end=""` — изглежда правилно, но задачата иска **число** M (при `120` ще излезе `021`).

## Задача 7 ⭐⭐ Просто ли е числото?
Въведи естествено число **N ≥ 2**. Отпечатай `Просто`, ако се дели само на 1 и на себе си, иначе `Не е просто`.

| Вход | Изход |
|---|---|
| `17` | `Просто` |
| `21` | `Не е просто` |

**Как мислим**
- Кога числото **не** е просто? Ако има делител между 2 и N − 1.
- Шаблон флаг: `has_divisor = False`; ако намерим делител → `True`. Отговорът — **след** цикъла.
- Защо не печатаме в цикъла? Ще отпечатаме „Просто“ много пъти!

**Пишем по стъпки**
1. Вход, `has_divisor = False`.
2. `for d in range(2, n):` → `if n % d == 0: has_divisor = True`.
3. След цикъла: `if has_divisor:` → `Не е просто`, `else` → `Просто`.
4. Бонус: `break` — щом намерим делител, няма смисъл да продължаваме.

````python
n = int(input("N: "))
has_divisor = False
for d in range(2, n):
    if n % d == 0:
        has_divisor = True
        break
if has_divisor:
    print("Не е просто")
else:
    print("Просто")
````

**Проверка:** `17` → `Просто` · `21` → `Не е просто` · `2` → `Просто` (цикълът е празен!)

**За бързите:** достатъчно е да проверим до √N: `range(2, int(n ** 0.5) + 1)`. Защо? Ако 36 = 4 · 9, единият делител винаги е ≤ 6.

## Задача 8 ⭐⭐⭐ Най-голямата цифра и колко пъти се среща
Въведи естествено число **N**. Отпечатай най-голямата му цифра и колко пъти се среща.

| Вход | Изход |
|---|---|
| `4342` | `4 2` |
| `909` | `9 2` |

**Как мислим**
- Шаблони максимум + брояч, но **заедно** и по цифрите.
- Три случая за всяка цифра: по-голяма от досегашната → нов максимум, броячът **става 1**; равна → броячът **+1**; по-малка → нищо.
- С какво започва `best`? С -1 (по-малко от всяка цифра) — така първата цифра винаги става максимум.

**Пишем по стъпки**
1. Вход, `best = -1`, `count = 0`.
2. Цикълът за цифри с `digit = n % 10`.
3. `if digit > best:` → `best = digit`, `count = 1`; `elif digit == best:` → `count += 1`.
4. `print(best, count)`.

````python
n = int(input("N: "))
best = -1
count = 0
while n > 0:
    digit = n % 10
    if digit > best:
        best = digit
        count = 1
    elif digit == best:
        count += 1
    n = n // 10
print(best, count)
````

**Проверка:** `4342` → `4 2` · `909` → `9 2` · `5` → `5 1`

Проследете `4342` с таблица (цифрите идват 2, 4, 3, 4): best 2/1 → 4/1 → 4/1 → 4/2.

**Чести грешки:** `count = 0` вместо `count = 1` при нов максимум.

## Задача 9 ⭐⭐⭐ Две съседни нули
Въведи естествено число **N**. Отпечатай `Да`, ако в записа му има две нули една до друга, иначе `Не`.

| Вход | Изход |
|---|---|
| `1004` | `Да` |
| `1030` | `Не` |

**Как мислим**
- Гледаме цифрите една по една. Какво трябва да помним, за да знаем дали две **съседни** са нули? **Предишната** цифра.
- В началото няма предишна → `prev = -1` (нещо, което не е 0).
- Флаг `found` → отговорът е след цикъла.

**Пишем по стъпки**
1. Вход, `prev = -1`, `found = False`.
2. Цикълът за цифри: `digit = n % 10`.
3. `if digit == 0 and prev == 0: found = True`.
4. **Накрая** в цикъла: `prev = digit` (текущата става предишна за следващата стъпка).
5. След цикъла: `Да` / `Не`.

````python
n = int(input("N: "))
prev = -1
found = False
while n > 0:
    digit = n % 10
    if digit == 0 and prev == 0:
        found = True
    prev = digit
    n = n // 10
if found:
    print("Да")
else:
    print("Не")
````

**Проверка:** `1004` → `Да` · `1030` → `Не` · `100` → `Да` · `505` → `Не`

**Чести грешки:** `prev = digit` **преди** проверката — тогава винаги сравняваме цифрата със самата нея.

## Задача 10 ⭐⭐⭐ Повтарящи се цифри
Въведи естествено число **N**. Отпечатай `Да`, ако в него има цифра, която се среща повече от веднъж, иначе `Не`.

| Вход | Изход |
|---|---|
| `1234` | `Не` |
| `1232` | `Да` |

**Как мислим**
- Можем да попитаме за всяка цифра от 0 до 9: колко пъти се среща в N? Ако някоя е ≥ 2 → Да.
- „Колко пъти се среща цифра“ = цикълът за цифри + брояч. Слагаме го **вътре** в `for c in range(10)` → цикъл в цикъл.
- Капан: вътрешният цикъл „изяжда“ числото (`n` става 0). Затова за всяка цифра работим с копие: `temp = n`.

**Пишем по стъпки**
1. Вход, `repeated = False`.
2. `for c in range(10):` → `temp = n`, `count = 0`.
3. Вътрешният `while temp > 0:` → `if temp % 10 == c: count += 1`, `temp = temp // 10`.
4. След вътрешния цикъл: `if count >= 2: repeated = True`.
5. Накрая: `Да` / `Не`.

````python
n = int(input("N: "))
repeated = False
for c in range(10):
    temp = n
    count = 0
    while temp > 0:
        if temp % 10 == c:
            count += 1
        temp = temp // 10
    if count >= 2:
        repeated = True
if repeated:
    print("Да")
else:
    print("Не")
````

**Проверка:** `1234` → `Не` · `1232` → `Да` · `100` → `Да` (нулата два пъти) · `7` → `Не`

**Чести грешки:** без `temp = n` — след проверката за цифра 0 числото е станало 0 и за другите цифри не се брои нищо.
