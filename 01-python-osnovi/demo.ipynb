## 1. `print` и коментари

Програмата е списък с инструкции, които компютърът изпълнява **отгоре надолу**.
`print()` показва нещо на екрана. Всичко след `#` е коментар — Python го пропуска.

```python
print("Здравей, ТУЕС!")
print("Python е готин")  # коментар: това не се изпълнява
print(2026)
```

```python
# Няколко неща наведнъж — разделени със запетая
print("Имам", 3, "котки")
```

## 2. Променливи и типове

Променливата е **име**, което сочи към стойност. Като етикет на кутия.
`=` означава „запомни“, а не „равно“.

```python
name = "Мария"
age = 14
height = 1.62
likes_python = True

print(name, age, height, likes_python)
```

```python
# Всяка стойност има тип
print(type(name))          # str   — текст
print(type(age))           # int   — цяло число
print(type(height))        # float — дробно число
print(type(likes_python))  # bool  — True / False
```

```python
# Променливата може да се презапише
age = 14
age = age + 1     # вземи старата стойност, добави 1, запиши обратно
print(age)

age += 1          # същото, по-кратко
print(age)
```

**f-string** — най-удобният начин да вмъкнеш променливи в текст. Сложи `f` пред кавичките и променливите в `{}`.

```python
print(f"Казвам се {name} и съм на {age} години.")
print(f"След 5 години ще съм на {age + 5}.")
```

**Правила за имена:** букви, цифри и `_`; не започват с цифра; без интервали. Главните и малките букви са различни: `age` ≠ `Age`.

## 3. Аритметика

| Оператор | Значение | Пример | Резултат |
|---|---|---|---|
| `+` `-` `*` | събиране, изваждане, умножение | `7 * 3` | `21` |
| `/` | деление (винаги `float`) | `7 / 2` | `3.5` |
| `//` | целочислено деление | `7 // 2` | `3` |
| `%` | остатък | `7 % 2` | `1` |
| `**` | степен | `2 ** 10` | `1024` |

```python
print(7 / 2)
print(7 // 2)
print(7 % 2)
print(2 ** 10)
```

```python
# Къде ни трябват // и % ? Например: 135 минути = ? часа и ? минути
minutes = 135
hours = minutes // 60
rest = minutes % 60
print(f"{minutes} мин = {hours} ч и {rest} мин")
```

```python
# Редът на действията е като в математиката. Скобите помагат.
print(2 + 3 * 4)
print((2 + 3) * 4)
print(round(10 / 3, 2))  # закръгляне до 2 знака
```

## 4. Вход от потребителя

`input()` спира програмата и чака потребителят да напише нещо.
**Важно:** `input()` винаги връща **текст** (`str`), дори да напишеш число.

```python
user = input("Как се казваш? ")
print(f"Приятно ми е, {user}!")
```

⚠️ **Нарочна грешка** — опитваме да съберем текст и число.

```python
year = input("Коя година си роден? ")
print(2026 - year)
```

## 5. Сравнения и `bool`

Сравнението връща `True` или `False`.

`==` равно · `!=` различно · `<` `>` `<=` `>=`

⚠️ `=` записва стойност, `==` сравнява!

```python
print(5 > 3)
print(5 == 3)
print(5 != 3)
print("ябълка" == "Ябълка")   # главните букви имат значение
```

## 6. Условия: `if / elif / else`

Програмата избира какво да направи. Кодът вътре в `if` е **отместен** с 4 интервала (Tab) — така Python знае кое е „вътре“.

```python
temperature = 28

if temperature > 25:
    print("Горещо е 🥵")
    print("Вземи вода")
elif temperature > 15:
    print("Приятно е 🙂")
else:
    print("Студено е 🥶")

print("Това се изпълнява винаги")  # вече не е отместено
```

⚠️ **Нарочна грешка** — липсва отстъп.

```python
if temperature > 25:
print("Горещо е")
```

**Комбиниране на условия:** `and` (и двете), `or` (поне едното), `not` (обратното).

```python
age = 15
has_ticket = True

if age >= 12 and has_ticket:
    print("Заповядай в киното 🎬")
else:
    print("Съжалявам, не може")
```

```python
day = "събота"
if day == "събота" or day == "неделя":
    print("Почивен ден!")

is_raining = False
if not is_raining:
    print("Да излизаме навън")
```

## 7. Цикъл `for` и `range`

Цикълът повтаря код. `range(start, stop, step)` дава поредица от числа — **`stop` не се включва**.

```python
for i in range(5):
    print("Повторение", i)
```

```python
print(list(range(5)))          # 0..4
print(list(range(1, 6)))       # 1..5
print(list(range(0, 20, 5)))   # стъпка 5
print(list(range(10, 0, -1)))  # обратно броене
```

```python
# Цикъл + условие: четните числа до 20
for n in range(1, 21):
    if n % 2 == 0:
        print(n, end=" ")
```

```python
# Цикъл и по текст
for letter in "ТУЕС":
    print(letter)
```

## 8. Цикъл `while`

`while` повтаря, **докато** условието е вярно. Използваме го, когато не знаем предварително колко пъти.

⚠️ Ако условието никога не стане `False` → безкраен цикъл. В Colab го спираш с ■ (Stop).

```python
countdown = 5
while countdown > 0:
    print(countdown)
    countdown -= 1      # без този ред цикълът никога няма да спре!
print("Старт! 🚀")
```

```python
# break — излиза веднага от цикъла
while True:
    word = input("Напиши дума (или 'стоп'): ")
    if word == "стоп":
        break
    print(f"Ти написа: {word}")
print("Край")
```

## 9. Натрупване: сума и брояч

Много задачи се решават така: започваш от 0 и в цикъла **добавяш** по малко.

```python
total = 0
for n in range(1, 101):
    total += n
print("Сборът от 1 до 100 е", total)
```

```python
count = 0
for n in range(1, 101):
    if n % 7 == 0:
        count += 1
print("Числата от 1 до 100, които се делят на 7:", count)
```

## 10. Списъци

Списъкът пази **много стойности** в една променлива. Индексите започват от **0**.

```python
fruits = ["ябълка", "круша", "банан"]

print(fruits[0])      # първият
print(fruits[-1])     # последният
print(len(fruits))    # колко са

fruits.append("киви") # добави в края
print(fruits)
```

```python
for fruit in fruits:
    print(f"Обичам {fruit}")
```

```python
grades = [6, 5, 4, 6, 3]
print("Сбор:", sum(grades))
print("Най-висока:", max(grades))
print("Средна:", round(sum(grades) / len(grades), 2))
```

⚠️ **Нарочна грешка** — индекс извън списъка.

```python
print(fruits[10])
```

## 11. Всичко заедно

Мини програма: въвеждаме оценки, докато не напишем 0, и накрая получаваме статистика.
Тук има всичко от днес: вход, `int()`, `while`, `if`, списък, натрупване и f-string.

```python
grades = []

while True:
    grade = int(input("Оценка (0 за край): "))
    if grade == 0:
        break
    if grade < 2 or grade > 6:
        print("Оценката трябва да е от 2 до 6!")
        continue          # пропусни останалото и започни отначало
    grades.append(grade)

if len(grades) > 0:
    average = sum(grades) / len(grades)
    print(f"Въведени оценки: {grades}")
    print(f"Средна оценка: {average:.2f}")
    if average >= 5.5:
        print("Отличен успех! 🏆")
else:
    print("Не въведе нито една оценка.")
```
